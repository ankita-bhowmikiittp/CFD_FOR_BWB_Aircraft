```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Grid Parameters
grid_size = 50       # Number of points along each axis (50x50x50)
radius = 15          # Radius of the sphere in grid units
center = grid_size // 2  # Center coordinates (x0, y0, z0)

# 2. Create empty 3D grid array
grid = np.zeros((grid_size, grid_size, grid_size))

# 3. Populate grid using simple loops
for x in range(grid_size):
    for y in range(grid_size):
        for z in range(grid_size):
            # Equation of a sphere: (x - x0)^2 + (y - y0)^2 + (z - z0)^2 <= R^2
            distance_squared = (x - center)**2 + (y - center)**2 + (z - center)**2
            
            if distance_squared <= radius**2:
                grid[x][y][z] = 1  # Point is inside the sphere

# 4. Plotting the 3D Sphere
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

# Render filled voxels where grid == 1
ax.voxels(grid, edgecolor='black', alpha=0.5, facecolors='cyan')

ax.set_title("3D Mesh Grid Sphere via Simple Loops")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```
