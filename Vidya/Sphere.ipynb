```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Sphere Parameters
R = 10
N = 100 # Grid resolution

# 2. Spherical Angle Grids
theta = np.linspace(0, 2 * np.pi, N)  # Azimuthal angle (0 to 2pi)
phi = np.linspace(0, np.pi, N)        # Polar angle (0 to pi)

# Create 2D meshgrid for angles
theta_grid, phi_grid = np.meshgrid(theta, phi)

# 3. Parametric Equations for Sphere Surface
x = R * np.sin(phi_grid) * np.cos(theta_grid)
y = R * np.sin(phi_grid) * np.sin(theta_grid)
z = R * np.cos(phi_grid)

# 4. Plotting
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

# Replace this line:
surface = ax.plot_surface(x, y, z, color='cyan', edgecolor='k', linewidth=0.2, alpha=0.9)

# With this:
surface = ax.plot_surface(x, y, z, color='cyan', edgecolor='k', linewidth=0.2, alpha=0.9)
# Formatting
ax.set_title(f"3D Parametric Sphere (Radius = {R})", fontsize=14)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

# Equal aspect ratio for proper sphere proportions
ax.set_box_aspect([1, 1, 1])

# plt.colorbar(surface, shrink=0.5, aspect=10)
plt.show()
```
